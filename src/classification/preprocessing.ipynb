```python

# prepering the data for classification

import pandas as pd

df = pd.read_csv("../../data/H2.csv")

print(df.shape)
print(df.head())
```

```python
print("Dataset Columns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
print(df[["IsCanceled"]].value_counts())
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Columns:")
for column in df.columns:
    print(column)
```

```python
y = df["IsCanceled"]
print("Target Distribution:")
print(y.value_counts())

print("\nTarget Percentage:")
print(y.value_counts(normalize=True) * 100)
```

```python
for i, column in enumerate(df.columns, start=1):
    print(i, "-", column)
```

```python
features = [
    "LeadTime",
    "ArrivalDateYear",
    "ArrivalDateMonth",
    "ArrivalDateWeekNumber",
    "ArrivalDateDayOfMonth",
    "StaysInWeekendNights",
    "StaysInWeekNights",
    "Adults",
    "Children",
    "Babies",
    "Meal",
    "Country",
    "MarketSegment",
    "DistributionChannel",
    "IsRepeatedGuest",
    "PreviousCancellations",
    "PreviousBookingsNotCanceled",
    "ReservedRoomType",
    "CustomerType"
]


X = df[features].copy()
```

```python
print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
print("Numerical Features:")
print(X.select_dtypes(include=["int64", "float64"]).columns.tolist())

print("\nCategorical Features:")
print(X.select_dtypes(include=["object"]).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical:", numeric_features)
print("Categorical:", categorical_features)
```

```python
print("\nMissing Values in Training Data:")
print(X_train.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer

num_imputer = SimpleImputer(strategy="median")

X_train[numeric_features] = num_imputer.fit_transform(
    X_train[numeric_features]
)

X_test[numeric_features] = num_imputer.transform(
    X_test[numeric_features]
)
```

```python
cat_imputer = SimpleImputer(strategy="most_frequent")

X_train[categorical_features] = cat_imputer.fit_transform(
    X_train[categorical_features]
)

X_test[categorical_features] = cat_imputer.transform(
    X_test[categorical_features]
)

print("Training Missing Values:", X_train.isnull().sum().sum())
print("Testing Missing Values:", X_test.isnull().sum().sum())

```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown="ignore")

X_train_encoded = encoder.fit_transform(
    X_train[categorical_features]
)

X_test_encoded = encoder.transform(
    X_test[categorical_features]
)

print("Original Shape:", X_train[categorical_features].shape)
print("Encoded Shape:", X_train_encoded.shape)

print("\nEncoded Feature Names:")
print(encoder.get_feature_names_out(categorical_features))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train[numeric_features]
)

X_test_scaled = scaler.transform(
    X_test[numeric_features]
)

print("Before Scaling:")
print(X_train[numeric_features].head())

print("\nAfter Scaling:")
print(X_train_scaled[:5])
```

```python
from scipy.sparse import hstack

X_train_final = hstack([
    X_train_scaled,
    X_train_encoded
]).tocsr()

X_test_final = hstack([
    X_test_scaled,
    X_test_encoded
]).tocsr()

print("Training Shape:", X_train_final.shape)
print("Testing Shape:", X_test_final.shape)

print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)
```

```python
encoded_features = encoder.get_feature_names_out(categorical_features)

all_features = list(numeric_features) + list(encoded_features)

X_train_df = pd.DataFrame(
    X_train_final.toarray(),
    columns=all_features
)

X_test_df = pd.DataFrame(
    X_test_final.toarray(),
    columns=all_features
)

train_df = X_train_df.copy()
train_df["IsCanceled"] = y_train.reset_index(drop=True)

test_df = X_test_df.copy()
test_df["IsCanceled"] = y_test.reset_index(drop=True)

train_df.to_csv("../../data/train_preprocessed.csv", index=False)

test_df.to_csv("../../data/test_preprocessed.csv", index=False)

print(train_df.shape)
print(test_df.shape)

print(train_df.head())
```
